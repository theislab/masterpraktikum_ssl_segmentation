```python
from pydeseq2.dds import DeseqDataSet
from pydeseq2.ds import DeseqStats

import pandas as pd
```

```python
counts = pd.read_csv('DEGs//EC_Intima_pb.csv', index_col=0)
counts
```

```python
metadata = pd.DataFrame(zip(counts.index, counts["condition"]),
                        columns = ['Sample', 'Condition'])
metadata = metadata.set_index('Sample')
metadata
```

```python
counts = counts.iloc[:,:-4]
counts
```

```python
dds = DeseqDataSet(counts=counts,
            metadata=metadata,
            design_factors="Condition")
```

```python
dds.deseq2()
```
